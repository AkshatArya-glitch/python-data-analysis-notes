```python
# from a pandas dataframe

batsman = pd.read_csv('sharma-kohli.csv')
batsman
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# plotting multiple plots 
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# labels and title
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.title("Virat Kohli vs Rohit Sharma: A Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs")
```

```python
# color(hex) and line(width and style) and maker(size)
plt.plot(batsman['index'], batsman['V Kohli'],color='#fc030f')
plt.plot(batsman['index'], batsman['RG Sharma'],color='#0f03fc')
plt.title("Virat Kohli vs Rohit Sharma: A Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.show()
```

```python
# line style ( dashdot, dashed, dotted ) linewidth=3 or 2 or 1
plt.plot(batsman['index'], batsman['V Kohli'],color='#fc030f',linestyle='dashdot',linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'],color='blue',linestyle='dashed',linewidth=3)
plt.title("Virat Kohli vs Rohit Sharma: A Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.show()
```

```python
# Marker diamond('D','O') markersize(5,20,10,3)
plt.plot(batsman['index'], batsman['V Kohli'],color='#fc030f',linestyle='dashdot',linewidth=3,marker='*',markersize=11)
plt.plot(batsman['index'], batsman['RG Sharma'],color='#0f03fc',linestyle='dashed',linewidth=3,marker='D',markersize=6)
plt.title("Virat Kohli vs Rohit Sharma: A Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.show()
```

```python
# legend-> location
plt.plot(batsman['index'], batsman['V Kohli'],color='#fc030f',linestyle='dashdot',linewidth=3,marker='*',markersize=11,label='V Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'],color='#0f03fc',linestyle='dashed',linewidth=3,marker='D',markersize=6,label='RG Sharma')
plt.title("Virat Kohli vs Rohit Sharma: A Career Comparison")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.legend(loc='best')
plt.show()
```

Bar Chart

Bivariate Analysis

Numerical vs Categorical 

Use Case- Aggregate analysis of groups

```python
# Simple bar chart 
children=[10,20,40,10,50]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color='pink')
```

```python
# horizontal bar using # colors and label 

plt.barh(colors,children,color='black')
```

```python
# colors and label 
df=pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
# multiple bar chart & xticks
# np.arange(df.shape[0])
# plt.bar(np.arange(df.shape[0]) -0.5,df['2015'])
plt.bar(np.arange(df.shape[0]) -0.2,df['2015'],width=0.2,color='black')
plt.bar(np.arange(df.shape[0]) ,df['2016'],width=0.2,color='grey')
plt.bar(np.arange(df.shape[0]) +0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# a problem
children=[10,20,40,10,50]
colors=['red blue green yellow pink', ' red blue black yellow pink', 'red blue green yellow green','red yellow green yellow pink', ' red blue black yellow pink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# stack bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016']+df['2017']),label='2015')
plt.legend()
plt.show()
```

Pie Chart

```python
# simple data 
data=[23,45,100,20,40]
subject=['Eng','Science','Maths','SST','Hindi']
plt.pie(data,labels=subject)
plt.show()
```

```python
# datset
df=pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'])
plt.show()
```

```python
# percentage and colors 
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['green','red','black','grey','brown','pink'])
plt.show()
```
