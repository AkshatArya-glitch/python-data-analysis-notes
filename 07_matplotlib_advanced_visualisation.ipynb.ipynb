```python
fig, axes=plt.subplots(nrows=2, ncols=1, sharex= True, figsize=(10,7))

axes[0].scatter(batter['avg'],batter['strike_rate'], color='red')
axes[1].scatter(batter['avg'],batter['runs'])

axes[0].set_title('Avg VS Strike Rate')
# axes[0].set_xlabel('Average')
axes[0].set_ylabel('Strike Rate')
axes[1].set_title('Avg VS Runs')

axes[1].set_title('Avg VS Runs')
axes[1].set_xlabel('Average')
axes[1].set_ylabel('Runs')

```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10,10))
axes[0,0].scatter(batter['avg'],batter['strike_rate'],color='red')
axes[0,1].scatter(batter['avg'],batter['runs'])
axes[1,0].hist(batter['avg'])
axes[1,1].hist(batter['runs'])
plt.show()
```

```python
fig=plt.figure()

ax1=fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

ax2=fig.add_subplot(2,2,2)
ax2.scatter(batter['avg'],batter['runs'])

ax3=fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])

ax4=fig.add_subplot(2,2,4)
ax4.hist(batter['runs'],color='green')

plt.show()
```

```python
fig=plt.figure()
ax=fig.add_subplot(projection='3d')

ax.scatter(batter['runs'],batter['avg'],batter['strike_rate'])
ax.set_title('3D Scatter Plot')
ax.set_xlabel('Runs')
ax.set_ylabel('Average')
ax.set_zlabel('Strike rate')
```

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)  # it create a grid of x and y values. It is used to create a 2D grid of points from the 1D arrays x and y. The resulting
# xx and yy arrays will have the same shape, where each element in xx corresponds to a value from x and each element in yy corresponds to a value 
# from y. This is often used for plotting functions or surfaces in 3D space.
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure(figsize=(12,8))
ax= fig.add_subplot( projection='3d')
p=ax.plot_surface(xx,yy,z, cmap='viridis')

plt.colorbar(p)
plt.show()
```

```python
z=np.sin(xx)+np.cos(yy)

fig=plt.figure(figsize=(12,8))
ax=fig.add_subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='jet')

plt.colorbar(p)
plt.show()
```

```python
# Contour plots ( it conerts the 3D surface into 2D plot with contour lines)
z=xx**2+yy**2
fig=plt.figure(figsize=(12,8))
ax= fig.add_subplot()

contour= ax.contourf(xx,yy,z,cmap='jet')
fig.colorbar(contour)
```

```python
z=np.sin(xx)+np.cos(yy)

fig=plt.figure(figsize=(12,8))
ax=fig.add_subplot()
p=ax.contourf(xx,yy,z,cmap='jet')
plt.colorbar(p)
plt.show()
```

```python
delivery =pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6]))&(delivery['batsman_run']==6)]
temp_df
```

```python
grid=temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
grid
```

```python
plt.figure(figsize=(20,6))
plt.imshow(grid,cmap='jet')
plt.yticks(delivery['overs'].unique(),list(range(1,21)))
plt.xticks(np.arange(0,6), list(range(1,7)))
plt.colorbar()
plt.show()
```
